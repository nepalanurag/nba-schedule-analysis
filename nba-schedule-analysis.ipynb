## Method: schedule effect on wins

**Question 9.** Estimate how many more/fewer regular-season wins each team has had due to schedule-related factors from 2019-20 through 2023-24. One number per team, total wins, not per-82.

**Model.** I fit a linear regression predicting win probability per game. Team strength enters as season-to-date offensive and defensive eFG% edges, computed leave-one-out and shrunk toward the league average. Schedule effects enter as rest days, back-to-backs, 3-in-4 and 4-in-6 flags, travel distance, cumulative travel, rest differential versus the opponent, and travel/rest interactions. The regression summary shows the shooting efficiency edges and home court dominate; schedule variables are weaker and less consistent. Then two counterfactuals: a neutral schedule (rest set to 2 days, no back-to-backs or 4-in-6s, median travel, equal rest) and a league-average schedule. The per-game win-probability differences versus each counterfactual, summed by team, give the wins gained or lost to the schedule.

**Answer 9:**
- Most helped by schedule: MIL (+0.71 wins)
- Most hurt by schedule: DEN (-1.03 wins)

## Takeaway

Schedule effects are real but small next to team quality. Shooting efficiency edges and home court dominate win probability, while rest and travel nibble at the margins. Over five seasons the luckiest schedule (MIL, +0.71) and the unluckiest (DEN, -1.03) differ by under two wins total. The decade trend is kinder scheduling: back-to-backs fell by a third from 2014 to 2019 before the pandemic season undid some of that progress.
